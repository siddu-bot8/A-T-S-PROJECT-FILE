```python
# 4. Count employees department-wise
print(emp['Department'].value_counts())

emp.groupby('Department')['EmpID'].count()
# the line Name: count, dtype: int64 is pandas telling you about the Series metadata:

# Name: count → the Series has been given the name "count".
# This usually happens when you run something like df['Department'].value_counts() or groupby().count(). 
# The operation labels the resulting Series with "count" because it represents counts.

# dtype: int64 → the data type of the values in the Series is int64 (NumPy’s integer type).
```

```python
# 1. Bar chart for department-wise employee count
dept_counts = emp['Department'].value_counts()
plt.bar(dept_counts.index, dept_counts.values)
plt.title("Department-wise Employee Count")
plt.show()
```
