# MATPLOTLIB

```python
df[['Maths', 'Science', 'English']].mean()
```

```python
avg_marks = df[['Maths', 'Science', 'English']].mean()
plt.bar(avg_marks.index, avg_marks.values)

plt.ylim(0, 100)

plt.title('Average Subject Marks')
plt.xlabel('Subjects')
plt.ylabel('Average Marks')

plt.show()
```

```python
bars = plt.bar(avg_marks.index, avg_marks.values)

plt.bar_label(bars)

plt.ylim(0, 100)

plt.title('Average Subject Marks')
plt.xlabel('Subjects')
plt.ylabel('Average Marks')

plt.show()
```

```python
plt.hist(df['Total'], bins=10)
plt.title('Histogram of Total Marks')
plt.xlabel('Total Marks')
plt.ylabel('Frequency')
plt.show()
```

```python
grade_counts = df['Grade'].value_counts()

plt.pie(grade_counts.values,
        labels=grade_counts.index,
        autopct='%1.1f%%')

plt.title('Pie Chart of Grades')
plt.show()
```
