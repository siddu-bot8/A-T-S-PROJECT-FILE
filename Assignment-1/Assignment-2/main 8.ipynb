# MATPLOTLIB

```python
membership_counts = df['MembershipType'].value_counts()

# Plot pie chart
plt.figure(figsize=(6,6))
plt.pie(membership_counts, labels=membership_counts.index, autopct='%1.1f%%', startangle=90)
plt.title("Membership Types Distribution")
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.hist(df['Age'], bins=5, color='skyblue', edgecolor='black')
plt.title("Age Distribution of Members")
plt.xlabel("Age")
plt.ylabel("Number of Members")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
trainer_counts = df['TrainerAssigned'].value_counts()

# Plot bar chart
plt.figure(figsize=(6,4))
trainer_counts.plot(kind='bar', color=['green','red'], edgecolor='black')
plt.title("Trainer Assignment Status")
plt.xlabel("Trainer Assigned")
plt.ylabel("Number of Members")
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
