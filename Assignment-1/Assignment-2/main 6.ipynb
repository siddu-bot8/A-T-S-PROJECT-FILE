# MATPLOTLIB

```python
import matplotlib.pyplot as plt

# Count loan statuses
loan_counts = df['LoanStatus'].value_counts()

# Plot bar chart
loan_counts.plot(kind='bar', color=['skyblue', 'lightgreen', 'salmon'], edgecolor='black')

# Add labels and title
plt.xlabel('Loan Status')
plt.ylabel('Number of Applicants')
plt.title('Loan Status Distribution')

# Add counts on top of bars
for i, count in enumerate(loan_counts.values):
    plt.text(i, count + 0.5, str(count), ha='center')

plt.show()
```

```python
import matplotlib.pyplot as plt

# Histogram of CreditScore column
plt.hist(df['CreditScore'], bins=10, edgecolor='black', color='skyblue')

# Add labels and title
plt.xlabel('Credit Score')
plt.ylabel('Number of Customers')
plt.title('Distribution of Credit Scores')

plt.show()
```

```python
import matplotlib.pyplot as plt

# Scatter plot
plt.scatter(df['Income'], df['LoanAmount'], color='skyblue', edgecolor='black')

# Add labels and title
plt.xlabel('Income')
plt.ylabel('Loan Amount')
plt.title('Income vs Loan Amount')

plt.show()
```
