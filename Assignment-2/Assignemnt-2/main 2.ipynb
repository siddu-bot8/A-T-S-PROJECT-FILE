```python
discount_counts = df['discount_percent'].value_counts().sort_index()
print(discount_counts)
```

```python
df['discount_percent'].mean()
```

```python
discount_update = df['discount_percent'].mean()
df['discount_percent'] = df['discount_percent'].fillna(discount_update)
```

```python
df['discount_percent'].value_counts().sort_index()
```

```python
df.info()
```

```python
# 1. Create the revenue column
df['revenue'] = df['quantity'] * df['unit_price'] * (1 - df['discount_percent'] / 100)
```

```python
df.sort_values(by='revenue', ascending=False).head(5)
```

```python
df.groupby('category')['revenue'].agg(
    Total_Revenue='sum',
    Mean='mean',
    STD='std'
)
```

```python
df.groupby('returned')['order_id'].count()
```

```python
# df['returned']=='yes'.count()

(df['returned'] == 'Yes').sum()
```

```python
return_stats = (
    df.groupby("category")
      .agg(
          total=("order_id", "count"),
          returned=("returned", lambda x: (x == "Yes").sum())
      )
)

return_stats["return_rate"] = (return_stats["returned"] / return_stats["total"]) * 100
print(return_stats.sort_values("return_rate", ascending=False))


# https://copilot.microsoft.com/shares/oLuW4gQayGMfzADMtLPnz
```

```python
aov_stats = df.groupby(df["customer_id"].duplicated(keep=False))["revenue"].mean()
print(aov_stats)
```

```python
# Flag repeat customers
df["is_repeat"] = df["customer_id"].duplicated(keep=False)

# Group and rename labels
aov_stats = (
    df.groupby(df["is_repeat"].map({True: "Repeat", False: "New"}))["revenue"].mean()
)

print(aov_stats)

print
```

```python
import matplotlib.pyplot as plt

# Count orders per category
order_counts = df["category"].value_counts()

# Plot pie chart
plt.figure(figsize=(6,6))
plt.pie(order_counts, labels=order_counts.index, autopct="%1.1f%%", startangle=90)
plt.title("Order Count by Category")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
sns.barplot(
    data=df,
    x="shipping_mode",
    y="revenue",
    hue="returned",   # color by returned status
    estimator="mean"  # average revenue
)
plt.title("Average Revenue by Shipping Mode (Returned vs Not Returned)")
plt.show()
```
