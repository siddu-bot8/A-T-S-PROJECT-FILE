```python
bins = [0, 25000, 50000, 75000, 100000, 125000, 150000, 175000, 200000]
labels = ["0-25K","25K-50K","50K-75K","75K-100K","100K-125K","125K-150K","150K-175K","175K-200K"]

df["Salary_Range"] = pd.cut(df["Monthly_Salary"], bins=bins, labels=labels, right=False)
```

```python
summary = (
    df.groupby("Salary_Range")
      .agg(
          Total_Employees=("Employee_ID", "count"),
          Attrition_Yes=("Attrition_Status", lambda x: (x == "Yes").sum())
      )
)

# Calculate attrition rate (%)
summary["Attrition_Rate"] = (summary["Attrition_Yes"] / summary["Total_Employees"]) * 100

# Sort by attrition rate
# summary = summary.sort_values(by="Attrition_Rate", ascending=False).round(2)

summary["Attrition_Rate"] = summary["Attrition_Rate"].apply(lambda x: f"{x:.2f}%")

print(summary.sort_values(by="Attrition_Rate",ascending=False))
```
